**Давайте добавим новый признак.**

```python
def age_category(age):
    if age < 30:
        return 1
    elif age < 55:
        return 2
    elif age >= 55:
        return 3
```

```python
age_categories = [age_category(age) for age in data.Age]
data['Age_category'] = age_categories
```

**Другой способ сделать это с `apply`.**

```python
data['Age_category'] = data['Age'].apply(age_category)
```

**2. Определите распределение функции `Pclass`. Теперь Для мужчин и женщин отдельно. Сколько людей из второго класса было на борту?**

**3. Каковы медиана и стандартное отклонение`Fare`?. Округлите до 2-х знаков после запятой.**
- медиана 14,45, стандартное отклонение 49,69

** 4. Правда ли, что средний возраст выживших людей выше, чем у пассажиров, которые в конечном итоге умерли? **
- Да
- Нет


```python
# Средний возраст выживших и погибших
mean_age_survived = data.loc[data['Survived'] == 1, 'Age'].mean()
mean_age_died = data.loc[data['Survived'] == 0, 'Age'].mean()

print(f"Средний возраст выживших: {mean_age_survived:.2f}")
print(f"Средний возраст погибших: {mean_age_died:.2f}")
print("Средний возраст выживших выше:", mean_age_survived > mean_age_died)
```

**5. Это правда, что пассажиры моложе 30 лет. выжили чаще, чем те, кому больше 60 лет. Каковы доли выживших людей среди молодых и пожилых людей?**
- 22,7% среди молодежи и 40,6% среди пожилых
- 40,6% среди молодежи и 22,7% среди пожилых
- 35,3% среди молодежи и 27,4% среди пожилых
- 27,4% среди молодежи и 35,3% среди пожилых

```python
# Доля выживших среди пассажиров младше 30 и старше 60 лет
young = data[data['Age'] < 30]
old = data[data['Age'] > 60]

young_survival = young['Survived'].mean()
old_survival = old['Survived'].mean()

print(f"Младше 30 лет: {young_survival:.1%}")
print(f"Старше 60 лет: {old_survival:.1%}")
print("Молодые выживали чаще:", young_survival > old_survival)
```

**6. Правда ли, что женщины выживали чаще мужчин? Каковы доли выживших людей среди мужчин и женщин?**
- 30,2% среди мужчин и 46,2% среди женщин

**8. Как средний возраст мужчин / женщин зависит от `Pclass`? Выберите все правильные утверждения:**
- В среднем мужчины 1 класса старше 40 лет
- В среднем женщины 1 класса старше 40 лет
- Мужчины всех классов в среднем старше, чем женщины того же класса
- В среднем, пассажиры первого класса старше, чем пассажиры 2-го класса, которые старше, чем пассажиры 3-го класса.

```python
# Средний возраст мужчин и женщин в зависимости от класса
mean_age_by_class_sex = data.pivot_table(
    values='Age',
    index='Pclass',
    columns='Sex',
    aggfunc='mean'
)

print(mean_age_by_class_sex)

print("\nПроверка утверждений:")
print("1. Мужчины 1 класса старше 40 лет:",
      mean_age_by_class_sex.loc[1, 'male'] > 40)
print("2. Женщины 1 класса старше 40 лет:",
      mean_age_by_class_sex.loc[1, 'female'] > 40)
print("3. Мужчины всех классов старше женщин того же класса:",
      (mean_age_by_class_sex['male'] > mean_age_by_class_sex['female']).all())

class_means = data.groupby('Pclass')['Age'].mean()
print("4. Средний возраст: 1 класс > 2 класс > 3 класс:",
      class_means.loc[1] > class_means.loc[2] > class_means.loc[3])
```
